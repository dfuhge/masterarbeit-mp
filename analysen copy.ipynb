### Analyse Q.3-2 Korrespondenzanalyse

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt


# ============================================================
# 1. READ DATA
# ============================================================

df = pd.read_csv(
    "Schritte_final_ohneDoppel.csv",
    delimiter=";"
)

# ============================================================
# 2. CREATE CONTINGENCY TABLES
# ============================================================
#
# Double annotations are removed separately for each pair.
# ============================================================

# ------------------------------------------------------------
# 2.1 K x AE
# ------------------------------------------------------------

df_K_AE = df[
    ~df["K"].str.contains(",", na=False)
    &
    ~df["AE"].str.contains(",", na=False)
].copy()

m_K_AE = pd.crosstab(
    df_K_AE["K"],
    df_K_AE["AE"]
)

print("\n" + "=" * 70)
print("MATRIX K x AE")
print("=" * 70)
print(m_K_AE)


# ------------------------------------------------------------
# 2.2 K x AM
# ------------------------------------------------------------

df_K_AM = df[
    ~df["K"].str.contains(",", na=False)
    &
    ~df["AM"].str.contains(",", na=False)
].copy()

m_K_AM = pd.crosstab(
    df_K_AM["K"],
    df_K_AM["AM"]
)

print("\n" + "=" * 70)
print("MATRIX K x AM")
print("=" * 70)
print(m_K_AM)


# ------------------------------------------------------------
# 2.3 AE x AM
# ------------------------------------------------------------

df_AE_AM = df[
    ~df["AE"].str.contains(",", na=False)
    &
    ~df["AM"].str.contains(",", na=False)
].copy()

m_AE_AM = pd.crosstab(
    df_AE_AM["AE"],
    df_AE_AM["AM"]
)

print("\n" + "=" * 70)
print("MATRIX AE x AM")
print("=" * 70)
print(m_AE_AM)


# ============================================================
# 3. FUNCTION FOR AUTOMATIC LABEL POSITIONING
# ============================================================

def calculate_label_positions(
    points,
    x_range,
    y_range,
    iterations=400
):
    """
    Moves labels away from each other while keeping them close
    to the actual CA coordinates.

    Only the label positions are changed.
    The arrow endpoints remain unchanged.
    """

    # --------------------------------------------------------
    # Initial displacement away from origin
    # --------------------------------------------------------

    for point in points:
        x = point["x"]
        y = point["y"]
        length = np.sqrt(x ** 2 + y ** 2)
        if length == 0:
            direction_x = 1
            direction_y = 1
        else:
            direction_x = x / length
            direction_y = y / length
        point["label_x"] = (
            x
            + direction_x
            * x_range
            * 0.025
        )
        point["label_y"] = (
            y
            + direction_y
            * y_range
            * 0.025
        )


    # --------------------------------------------------------
    # Minimum approximate distance between labels
    # --------------------------------------------------------

    min_dx = x_range * 0.08
    min_dy = y_range * 0.045


    # --------------------------------------------------------
    # Iterative collision resolution
    # --------------------------------------------------------

    for _ in range(iterations):

        moved = False


        for i in range(len(points)):

            for j in range(
                i + 1,
                len(points)
            ):

                p1 = points[i]
                p2 = points[j]


                dx = (
                    p2["label_x"]
                    - p1["label_x"]
                )

                dy = (
                    p2["label_y"]
                    - p1["label_y"]
                )


                # --------------------------------------------
                # Detect overlap
                # --------------------------------------------

                if (
                    abs(dx) < min_dx
                    and
                    abs(dy) < min_dy
                ):

                    moved = True


                    if abs(dx) < 1e-8:

                        dx = (
                            1
                            if j % 2 == 0
                            else -1
                        ) * min_dx * 0.1


                    if abs(dy) < 1e-8:

                        dy = (
                            1
                            if j % 2 == 0
                            else -1
                        ) * min_dy * 0.1


                    push_x = (
                        min_dx
                        - abs(dx)
                    ) * 0.08

                    push_y = (
                        min_dy
                        - abs(dy)
                    ) * 0.08


                    sign_x = (
                        1
                        if dx >= 0
                        else -1
                    )

                    sign_y = (
                        1
                        if dy >= 0
                        else -1
                    )


                    p1["label_x"] -= (
                        sign_x
                        * push_x
                    )

                    p2["label_x"] += (
                        sign_x
                        * push_x
                    )


                    p1["label_y"] -= (
                        sign_y
                        * push_y
                    )

                    p2["label_y"] += (
                        sign_y
                        * push_y
                    )


        # ----------------------------------------------------
        # Slightly pull labels back toward arrow endpoints
        # ----------------------------------------------------

        for point in points:

            point["label_x"] += (
                point["x"]
                - point["label_x"]
            ) * 0.003

            point["label_y"] += (
                point["y"]
                - point["label_y"]
            ) * 0.003


        if not moved:
            break


    return points


# ============================================================
# 4. FUNCTION FOR CORRESPONDENCE ANALYSIS
# ============================================================

def correspondence_analysis(
    contingency_table,
    title="Correspondence Analysis",
    row_label="Row categories",
    column_label="Column categories",
    row_color="green",
    column_color="orange"
):

    # ========================================================
    # 4.1 CONVERT TABLE TO NUMPY ARRAY
    # ========================================================

    N = contingency_table.to_numpy(
        dtype=float
    )

    n = N.sum()


    if n == 0:

        raise ValueError(
            "The contingency table contains no observations."
        )


    # ========================================================
    # 4.2 CORRESPONDENCE MATRIX
    # ========================================================

    P = (
        N / n
    )


    # ========================================================
    # 4.3 ROW AND COLUMN MASSES
    # ========================================================

    row_masses = (
        P.sum(axis=1)
    )

    column_masses = (
        P.sum(axis=0)
    )


    # ========================================================
    # 4.4 REMOVE EMPTY ROWS AND COLUMNS
    # ========================================================

    valid_rows = (
        row_masses > 0
    )

    valid_columns = (
        column_masses > 0
    )


    P = P[
        np.ix_(
            valid_rows,
            valid_columns
        )
    ]


    row_labels = (
        contingency_table.index[
            valid_rows
        ]
    )

    column_labels = (
        contingency_table.columns[
            valid_columns
        ]
    )


    row_masses = (
        P.sum(axis=1)
    )

    column_masses = (
        P.sum(axis=0)
    )


    # ========================================================
    # 4.5 EXPECTED VALUES UNDER INDEPENDENCE
    # ========================================================

    expected = np.outer(
        row_masses,
        column_masses
    )


    # ========================================================
    # 4.6 STANDARDIZED RESIDUAL MATRIX
    # ========================================================

    Dr_inv_sqrt = np.diag(
        1 / np.sqrt(
            row_masses
        )
    )

    Dc_inv_sqrt = np.diag(
        1 / np.sqrt(
            column_masses
        )
    )


    S = (
        Dr_inv_sqrt
        @ (P - expected)
        @ Dc_inv_sqrt
    )


    # ========================================================
    # 4.7 SINGULAR VALUE DECOMPOSITION
    # ========================================================

    U, singular_values, VT = (
        np.linalg.svd(
            S,
            full_matrices=False
        )
    )

    V = (
        VT.T
    )


    # ========================================================
    # 4.8 EIGENVALUES / INERTIA
    # ========================================================

    eigenvalues = (
        singular_values ** 2
    )

    total_inertia = (
        eigenvalues.sum()
    )


    if total_inertia > 0:

        explained_inertia = (
            eigenvalues
            / total_inertia
            * 100
        )

    else:

        explained_inertia = (
            np.zeros_like(
                eigenvalues
            )
        )


    cumulative_inertia = (
        np.cumsum(
            explained_inertia
        )
    )


    # ========================================================
    # 4.9 PRINCIPAL COORDINATES
    # ========================================================

    row_coordinates = (
        Dr_inv_sqrt
        @ U
        @ np.diag(
            singular_values
        )
    )


    column_coordinates = (
        Dc_inv_sqrt
        @ V
        @ np.diag(
            singular_values
        )
    )


    # ========================================================
    # 4.10 CREATE DATAFRAMES
    # ========================================================

    row_coordinates_df = pd.DataFrame(
        row_coordinates,
        index=row_labels,
        columns=[
            f"Dim{i + 1}"
            for i in range(
                row_coordinates.shape[1]
            )
        ]
    )


    column_coordinates_df = pd.DataFrame(
        column_coordinates,
        index=column_labels,
        columns=[
            f"Dim{i + 1}"
            for i in range(
                column_coordinates.shape[1]
            )
        ]
    )


    # ========================================================
    # 4.11 INERTIA TABLE
    # ========================================================

    inertia_table = pd.DataFrame({
        "Eigenvalue":
            eigenvalues,

        "Explained inertia (%)":
            explained_inertia,

        "Cumulative inertia (%)":
            cumulative_inertia
    })


    inertia_table.index = [
        f"Dimension {i + 1}"
        for i in range(
            len(
                inertia_table
            )
        )
    ]


    # ========================================================
    # 5. PRINT RESULTS
    # ========================================================

    print("\n")
    print("=" * 70)
    print(title.upper())
    print("=" * 70)


    print(
        f"\nTotal observations: "
        f"{int(n)}"
    )


    print(
        f"\nTotal inertia: "
        f"{total_inertia:.4f}"
    )


    print(
        "\nDimensions:"
    )

    print(
        inertia_table.round(3)
    )


    print(
        f"\nCategories in {row_label}:"
    )

    print(
        list(
            row_coordinates_df.index
        )
    )


    print(
        f"\nCategories in {column_label}:"
    )

    print(
        list(
            column_coordinates_df.index
        )
    )


    print(
        f"\nCoordinates {row_label}:"
    )

    print(
        row_coordinates_df.iloc[
            :, :2
        ].round(3)
    )


    print(
        f"\nCoordinates {column_label}:"
    )

    print(
        column_coordinates_df.iloc[
            :, :2
        ].round(3)
    )


    # ========================================================
    # 6. PLOT FIRST TWO DIMENSIONS
    # ========================================================

    if (
        row_coordinates.shape[1] >= 2
        and
        column_coordinates.shape[1] >= 2
    ):

        fig, ax = plt.subplots(
            figsize=(10, 10)
        )


        # ====================================================
        # 6.1 COORDINATE VALUES
        # ====================================================

        all_x = np.concatenate([
            row_coordinates_df[
                "Dim1"
            ].values,
            column_coordinates_df[
                "Dim1"
            ].values
        ])


        all_y = np.concatenate([
            row_coordinates_df[
                "Dim2"
            ].values,
            column_coordinates_df[
                "Dim2"
            ].values
        ])


        # ====================================================
        # 6.2 DETERMINE COMMON SCALE
        # ====================================================

        max_coordinate = max(
            np.max(
                np.abs(
                    all_x
                )
            ),
            np.max(
                np.abs(
                    all_y
                )
            )
        )


        if max_coordinate == 0:
            max_coordinate = 1


        coordinate_range = (
            2 * max_coordinate
        )


        # ====================================================
        # 6.3 CREATE CATEGORY POINTS
        # ====================================================

        points = []


        for label, row in (
            row_coordinates_df.iterrows()
        ):

            points.append({
                "label": str(label),
                "x": row["Dim1"],
                "y": row["Dim2"],
                "color": row_color
            })


        for label, row in (
            column_coordinates_df.iterrows()
        ):

            points.append({
                "label": str(label),
                "x": row["Dim1"],
                "y": row["Dim2"],
                "color": column_color
            })


        # ====================================================
        # 6.4 CALCULATE NON-OVERLAPPING LABEL POSITIONS
        # ====================================================

        points = calculate_label_positions(
            points,
            x_range=coordinate_range,
            y_range=coordinate_range,
            iterations=500
        )


        # ====================================================
        # 6.5 DRAW CATEGORY ARROWS
        # ====================================================

        for point in points:

            ax.annotate(
                "",
                xy=(
                    point["x"],
                    point["y"]
                ),
                xytext=(
                    0,
                    0
                ),
                arrowprops=dict(
                    arrowstyle="->",
                    linewidth=2,
                    color=point["color"],
                    shrinkA=0,
                    shrinkB=0
                ),
                zorder=2
            )


        # ====================================================
        # 6.6 DRAW LABELS AND CONNECTOR LINES
        # ====================================================

        for point in points:

            # Connector line
            ax.plot(
                [
                    point["x"],
                    point["label_x"]
                ],
                [
                    point["y"],
                    point["label_y"]
                ],
                linewidth=0.6,
                color=point["color"],
                alpha=0.5,
                zorder=2
            )


            # Label
            ax.text(
                point["label_x"],
                point["label_y"],
                point["label"],
                fontsize=11,
                fontweight="bold",
                color=point["color"],
                horizontalalignment="center",
                verticalalignment="center",
                bbox=dict(
                    boxstyle="round,pad=0.16",
                    facecolor="white",
                    edgecolor="none",
                    alpha=0.9
                ),
                zorder=4
            )


        # ====================================================
        # 6.7 DETERMINE FINAL COMMON LIMIT
        # ====================================================

        label_x_values = np.array([
            p["label_x"]
            for p in points
        ])

        label_y_values = np.array([
            p["label_y"]
            for p in points
        ])


        complete_values = np.concatenate([
            np.abs(
                all_x
            ),
            np.abs(
                all_y
            ),
            np.abs(
                label_x_values
            ),
            np.abs(
                label_y_values
            )
        ])


        max_limit = (
            complete_values.max()
            * 1.10
        )


        if max_limit == 0:
            max_limit = 1


        # ====================================================
        # 6.8 SYMMETRICAL AND IDENTICAL AXIS LIMITS
        # ====================================================

        ax.set_xlim(
            -max_limit,
            max_limit
        )

        ax.set_ylim(
            -max_limit,
            max_limit
        )


        # ====================================================
        # 6.9 REFERENCE LINES
        # ====================================================

        ax.axhline(
            y=0,
            linewidth=0.8,
            color="black",
            alpha=0.5,
            zorder=1
        )

        ax.axvline(
            x=0,
            linewidth=0.8,
            color="black",
            alpha=0.5,
            zorder=1
        )


        # ====================================================
        # 6.10 AXIS LABELS
        # ====================================================

        ax.set_xlabel(
            "Dimension 1 "
            f"({explained_inertia[0]:.1f}% Inertia)",
            fontsize=12
        )


        ax.set_ylabel(
            "Dimension 2 "
            f"({explained_inertia[1]:.1f}% Inertia)",
            fontsize=12
        )


        # ====================================================
        # 6.11 TITLE
        # ====================================================

        ax.set_title(
            title,
            fontsize=14
        )


        # ====================================================
        # 6.12 GRID
        # ====================================================

        ax.grid(
            alpha=0.12
        )


        # ====================================================
        # 6.13 EQUAL SCALE
        # ====================================================

        ax.set_aspect(
            "equal",
            adjustable="box"
        )


        # ====================================================
        # 6.14 CLEAN LAYOUT
        # ====================================================

        plt.tight_layout()

        plt.show()


    else:

        print(
            "\nA two-dimensional plot cannot be created "
            "because fewer than two dimensions are available."
        )


    # ========================================================
    # 7. RETURN RESULTS
    # ========================================================

    return (
        row_coordinates_df,
        column_coordinates_df,
        inertia_table
    )


# ============================================================
# 8. CORRESPONDENCE ANALYSIS: K x AE
# ============================================================

rows_K_AE, cols_K_AE, inertia_K_AE = (
    correspondence_analysis(
        m_K_AE,
        title="Korrespondenzanalyse K × AE",
        row_label="K",
        column_label="AE",
        row_color="green",
        column_color="orange"
    )
)


# ============================================================
# 9. CORRESPONDENCE ANALYSIS: K x AM
# ============================================================

rows_K_AM, cols_K_AM, inertia_K_AM = (
    correspondence_analysis(
        m_K_AM,
        title="Korrespondenzanalyse K × AM",
        row_label="K",
        column_label="AM",
        row_color="green",
        column_color="blue"
    )
)


# ============================================================
# 10. CORRESPONDENCE ANALYSIS: AE x AM
# ============================================================

rows_AE_AM, cols_AE_AM, inertia_AE_AM = (
    correspondence_analysis(
        m_AE_AM,
        title="Korrespondenzanalyse AE × AM",
        row_label="AE",
        column_label="AM",
        row_color="orange",
        column_color="blue"
    )
)
```

### Analyse Q.3-3 Chi^2 Test

```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

from scipy.stats import chi2_contingency


# ============================================================
# Hilfsfunktion: Chi²-Test und Cramérs V
# ============================================================

def contingency_test(matrix, name):
    """
    Führt einen Chi²-Unabhängigkeitstest auf einer absoluten
    Kontingenzmatrix durch und berechnet Cramérs V.

    matrix: pandas.DataFrame mit absoluten Häufigkeiten
    name:   Bezeichnung der Analyse
    """

    # Für den Chi²-Test dürfen keine vollständig leeren
    # Zeilen oder Spalten enthalten sein
    test_matrix = matrix.loc[
        matrix.sum(axis=1) > 0,
        matrix.sum(axis=0) > 0
    ]

    # Chi²-Test
    chi2, p, dof, expected = chi2_contingency(
        test_matrix,
        correction=False
    )

    n = test_matrix.to_numpy().sum()
    r, c = test_matrix.shape

    # Cramérs V
    cramers_v = np.sqrt(
        chi2 / (n * min(r - 1, c - 1))
    )

    expected_df = pd.DataFrame(
        expected,
        index=test_matrix.index,
        columns=test_matrix.columns
    )

    # Kontrolle der erwarteten Zellhäufigkeiten
    n_expected_below_5 = (expected_df < 5).sum().sum()
    share_expected_below_5 = (
        n_expected_below_5 / expected_df.size * 100
    )

    min_expected = expected_df.min().min()

    result = {
        "Vergleich": name,
        "Chi2": chi2,
        "df": dof,
        "p": p,
        "Cramers_V": cramers_v,
        "N": n,
        "Min_expected": min_expected,
        "Expected_below_5": n_expected_below_5,
        "Share_expected_below_5": share_expected_below_5
    }

    return result, expected_df


# ============================================================
# Hilfsfunktion: Holm-Korrektur
# ============================================================

def holm_adjust(p_values):
    """
    Holm-Bonferroni-Korrektur für multiple Tests.
    Gibt adjustierte p-Werte in ursprünglicher Reihenfolge zurück.
    """

    p_values = np.asarray(p_values, dtype=float)
    m = len(p_values)

    order = np.argsort(p_values)
    sorted_p = p_values[order]

    adjusted_sorted = np.empty(m)

    for i, p in enumerate(sorted_p):
        adjusted_sorted[i] = (m - i) * p

    # Holm-adjustierte Werte müssen monoton nicht fallend sein
    adjusted_sorted = np.maximum.accumulate(adjusted_sorted)

    # Maximal 1
    adjusted_sorted = np.minimum(adjusted_sorted, 1.0)

    adjusted = np.empty(m)
    adjusted[order] = adjusted_sorted

    return adjusted


# ============================================================
# Hilfsfunktion: Prozentmatrix + Heatmap
# ============================================================

def plot_contingency(matrix, x_label, y_label, title):
    # Zeilenweise Prozentwerte
    matrix_p = (
        matrix
        .div(matrix.sum(axis=1).replace(0, np.nan), axis=0)
        .mul(100)
        .fillna(0)
    )

    print(f"\n{title}")
    print("Prozentuale Kontingenzmatrix:")
    print(matrix_p.round(1))

    plt.figure(figsize=(12, 8))

    sns.heatmap(
        matrix_p,
        annot=True,
        fmt=".1f",
        cmap="Blues",
        linewidths=0.5,
        cbar_kws={"label": "Anteil (%)"}
    )

    plt.title(title)
    plt.xlabel(x_label)
    plt.ylabel(y_label)

    plt.tight_layout()
    plt.show()

    return matrix_p


# ============================================================
# 1. Deskriptive Kontingenzmatrizen und Heatmaps
# ============================================================

#m_K_AE_p = plot_contingency(
#    m_K_AE,
#    x_label="AE",
#    y_label="K",
#    title="Prozentuale Kontingenzmatrix K × AE"
#)

#m_K_AM_p = plot_contingency(
#    m_K_AM,
#    x_label="AM",
#    y_label="K",
#    title="Prozentuale Kontingenzmatrix K × AM"
#)

#m_AE_AM_p = plot_contingency(
#    m_AE_AM,
#    x_label="AM",
#    y_label="AE",
#    title="Prozentuale Kontingenzmatrix AE × AM"
#)


# ============================================================
# 2. Chi²-Tests und Cramérs V
# ============================================================

results = []
expected_matrices = {}

result, expected = contingency_test(
    m_K_AE,
    "K × AE"
)
results.append(result)
expected_matrices["K × AE"] = expected

result, expected = contingency_test(
    m_K_AM,
    "K × AM"
)
results.append(result)
expected_matrices["K × AM"] = expected

result, expected = contingency_test(
    m_AE_AM,
    "AE × AM"
)
results.append(result)
expected_matrices["AE × AM"] = expected


# ============================================================
# 3. Holm-Korrektur für die drei Signifikanztests
# ============================================================

p_values = [r["p"] for r in results]
p_adjusted = holm_adjust(p_values)

for result, p_adj in zip(results, p_adjusted):
    result["p_Holm"] = p_adj
    result["Signifikant_05"] = p_adj < 0.05


# ============================================================
# 4. Ergebnistabelle
# ============================================================

results_df = pd.DataFrame(results)

print("\n\n==============================================")
print("ERGEBNISSE DER KONTINGENZANALYSEN")
print("==============================================\n")

print(
    results_df[
        [
            "Vergleich",
            "N",
            "Chi2",
            "df",
            "p",
            "p_Holm",
            "Cramers_V",
            "Min_expected",
            "Expected_below_5",
            "Share_expected_below_5",
            "Signifikant_05"
        ]
    ].round({
        "Chi2": 3,
        "p": 5,
        "p_Holm": 5,
        "Cramers_V": 3,
        "Min_expected": 2,
        "Share_expected_below_5": 1
    })
)


# ============================================================
# 5. Ausführliche Ausgabe je Vergleich
# ============================================================

for result in results:

    print("\n----------------------------------------------")
    print(result["Vergleich"])
    print("----------------------------------------------")

    print(f"N              = {result['N']}")
    print(f"Chi²           = {result['Chi2']:.3f}")
    print(f"df             = {result['df']}")
    print(f"p              = {result['p']:.6f}")
    print(f"p (Holm)       = {result['p_Holm']:.6f}")
    print(f"Cramérs V      = {result['Cramers_V']:.3f}")

    print(
        f"Erwartete Häufigkeiten < 5: "
        f"{result['Expected_below_5']} "
        f"({result['Share_expected_below_5']:.1f} %)"
    )

    print(
        f"Kleinste erwartete Häufigkeit: "
        f"{result['Min_expected']:.2f}"
    )

    if result["Signifikant_05"]:
        print(
            "Ergebnis: statistisch signifikanter Zusammenhang "
            "nach Holm-Korrektur."
        )
    else:
        print(
            "Ergebnis: kein statistisch signifikanter Zusammenhang "
            "nach Holm-Korrektur."
        )
```

### Analyse P.2-1: DFG Gesamt

```python
import pandas as pd
import pm4py
from collections import Counter


groups = ["K", "AE", "AM"]


# ============================================================
# 1. SETTINGS
# ============================================================

file_path = "Schritte_final.csv"

# Separator used for parallel activities
parallel_separator = ","

# Minimum frequency relative to the number of concepts.
#
# Example:
# min_support = 5
#
# If there are 100 concepts, a transition must occur at least
# 5 times to be shown.
min_support = 5


# ============================================================
# 2. READ CSV
# ============================================================

df_org = pd.read_csv(
    file_path,
    sep=";"
)


# ============================================================
# 3. HELPER FUNCTION FOR PARALLEL ACTIVITIES
# ============================================================

def parse_activities(value):
    """
    Converts a cell such as

        K-RB,K-RP

    into

        {"K-RB", "K-RP"}

    Activities within the same set are interpreted as occurring
    at the same process position. No transitions are created
    between them.
    """

    if pd.isna(value):
        return frozenset()

    activities = [
        activity.strip()
        for activity in str(value).split(parallel_separator)
        if activity.strip()
    ]

    # frozenset:
    # - removes accidental duplicates
    # - makes the order inside a parallel group irrelevant
    return frozenset(activities)


# ============================================================
# ANALYZE EACH CATEGORY GROUP
# ============================================================

for g in groups:

    print("\n\n")
    print("#" * 80)
    print(f"CATEGORY GROUP: {g}")
    print("#" * 80)


    # ============================================================
    # 4. KEEP RELEVANT COLUMNS
    # ============================================================

    df = df_org[
        [
            "Konzept-Nr.",
            "Schritt-Nr.",
            g
        ]
    ].copy()


    # ============================================================
    # 5. CLEAN DATA
    # ============================================================

    # Remove rows without an activity
    df = df.dropna(
        subset=[g]
    )


    # Extract numerical step number
    df["Schritt-Nr."] = pd.to_numeric(
        df["Schritt-Nr."]
        .astype(str)
        .str.extract(
            r"-(\d+)$",
            expand=False
        ),
        errors="coerce"
    )


    # Remove rows where step number could not be extracted
    df = df.dropna(
        subset=["Schritt-Nr."]
    )


    # Convert step number to integer
    df["Schritt-Nr."] = (
        df["Schritt-Nr."]
        .astype(int)
    )


    # Parse parallel activities
    df["_activities"] = (
        df[g]
        .apply(parse_activities)
    )


    # Remove empty activity groups
    df = df[
        df["_activities"].apply(len) > 0
    ]


    # Sort according to case and process step
    df = df.sort_values(
        by=[
            "Konzept-Nr.",
            "Schritt-Nr."
        ]
    ).reset_index(drop=True)


    # ============================================================
    # 6. COLLAPSE CONSECUTIVE REPETITIONS
    # ============================================================
    #
    # Repetitions are removed PER ACTIVITY.
    #
    # Example 1:
    #
    # A
    # A
    # B
    #
    # becomes:
    #
    # A
    # B
    #
    #
    # Example 2:
    #
    # A,B
    # A,C
    # D
    #
    # becomes:
    #
    # A,B
    # C
    # D
    #
    # because A occurs in two consecutive process positions.
    #
    #
    # Example 3:
    #
    # A,B
    # A,B
    # C
    #
    # becomes:
    #
    # A,B
    # C
    #
    #
    # Example 4:
    #
    # A
    # B
    # A
    #
    # remains:
    #
    # A
    # B
    # A
    #
    # because the second A is not consecutive.
    # ============================================================

    compressed_rows = []


    for case_id, case in df.groupby(
        "Konzept-Nr.",
        sort=False
    ):

        case = case.sort_values(
            "Schritt-Nr."
        )


        # Activities occurring in the immediately preceding
        # ORIGINAL process position
        previous_activities = frozenset()

        new_step = 1


        for _, row in case.iterrows():

            current_activities = row["_activities"]


            # Remove activities that already occurred in the
            # immediately preceding process position
            new_activities = (
                current_activities
                - previous_activities
            )


            # IMPORTANT:
            # Comparison for the next row must be made against
            # the original current group, not the compressed one.
            #
            # This ensures:
            #
            # A
            # A
            # A
            #
            # becomes only:
            #
            # A
            previous_activities = current_activities


            # If the complete process position consists only of
            # repetitions, it disappears after compression.
            if not new_activities:
                continue


            new_row = row.copy()

            # Renumber process positions after compression
            new_row["Schritt-Nr."] = new_step

            # Store compressed activity set
            new_row["_activities"] = frozenset(
                new_activities
            )

            # Human-readable version
            new_row[g] = ",".join(
                sorted(new_activities)
            )

            compressed_rows.append(
                new_row
            )

            new_step += 1


    # Create compressed dataframe
    df_compressed = pd.DataFrame(
        compressed_rows
    )


    # ============================================================
    # 7. CHECK RESULT
    # ============================================================

    print("\nOriginal number of process positions:")
    print(
        len(df)
    )

    print("\nOriginal number of individual activity occurrences:")
    print(
        df["_activities"]
        .apply(len)
        .sum()
    )


    if df_compressed.empty:

        print(
            "\nNo activities remain after compression."
        )

        continue


    print("\nNumber of process positions after compression:")
    print(
        len(df_compressed)
    )

    print("\nNumber of individual activity occurrences after compression:")
    print(
        df_compressed["_activities"]
        .apply(len)
        .sum()
    )

    print("\nNumber of concepts:")
    print(
        df_compressed["Konzept-Nr."]
        .nunique()
    )


    # ============================================================
    # 8. CREATE DFG MANUALLY
    # ============================================================
    #
    # Parallel activities are NOT ordered internally.
    #
    # Example:
    #
    # A
    # B,C
    # D
    #
    # produces:
    #
    # A -> B
    # A -> C
    # B -> D
    # C -> D
    #
    # but NOT:
    #
    # B -> C
    # C -> B
    #
    #
    # If two consecutive groups both contain several activities:
    #
    # A,B
    # C,D
    #
    # the Cartesian product is used:
    #
    # A -> C
    # A -> D
    # B -> C
    # B -> D
    # ============================================================

    dfg = Counter()

    start_activities = Counter()

    end_activities = Counter()


    for case_id, case in df_compressed.groupby(
        "Konzept-Nr.",
        sort=False
    ):

        case = case.sort_values(
            "Schritt-Nr."
        )


        activity_groups = (
            case["_activities"]
            .tolist()
        )


        if not activity_groups:
            continue


        # --------------------------------------------------------
        # Start activities
        # --------------------------------------------------------

        for activity in activity_groups[0]:

            start_activities[activity] += 1


        # --------------------------------------------------------
        # End activities
        # --------------------------------------------------------

        for activity in activity_groups[-1]:

            end_activities[activity] += 1


        # --------------------------------------------------------
        # Directly-follows relations
        # --------------------------------------------------------

        for previous_group, next_group in zip(
            activity_groups[:-1],
            activity_groups[1:]
        ):

            # Cartesian product between both process positions
            for source in previous_group:

                for target in next_group:

                    # Additional safeguard against self-loops
                    if source != target:

                        dfg[
                            (source, target)
                        ] += 1


    # Convert Counter objects to ordinary dictionaries
    dfg = dict(dfg)

    start_activities = dict(
        start_activities
    )

    end_activities = dict(
        end_activities
    )


    # ============================================================
    # 9. CALCULATE NUMBER OF CONCEPTS
    # ============================================================

    number_of_cases = (
        df_compressed[
            "Konzept-Nr."
        ]
        .nunique()
    )


    # ============================================================
    # 10. FILTER DFG BY TRANSITION FREQUENCY
    # ============================================================
    #
    # Example:
    #
    # number_of_cases = 100
    # min_support = 5
    #
    # => minimum frequency = 5
    #
    # A transition occurring fewer than 5 times is removed.
    #
    # IMPORTANT:
    # This is based on TRANSITION FREQUENCY, not on the number
    # of unique concepts containing the transition.
    #
    # A transition can therefore occur several times within
    # one concept.
    # ============================================================

    min_frequency = (
        number_of_cases
        * min_support
        / 100
    )


    dfg_filtered = {
        transition: frequency
        for transition, frequency
        in dfg.items()
        if frequency >= min_frequency
    }


    # ============================================================
    # 11. REMOVE SELF-REFERENCES AS ADDITIONAL SAFEGUARD
    # ============================================================

    dfg_filtered = {
        (source, target): frequency
        for (source, target), frequency
        in dfg_filtered.items()
        if source != target
    }


    # ============================================================
    # 12. KEEP ONLY VALID START/END ACTIVITIES
    # ============================================================

    activities_in_filtered_dfg = set()


    for source, target in dfg_filtered.keys():

        activities_in_filtered_dfg.add(
            source
        )

        activities_in_filtered_dfg.add(
            target
        )


    start_activities_filtered = {
        activity: frequency
        for activity, frequency
        in start_activities.items()
        if activity in activities_in_filtered_dfg
    }


    end_activities_filtered = {
        activity: frequency
        for activity, frequency
        in end_activities.items()
        if activity in activities_in_filtered_dfg
    }


    # ============================================================
    # 13. PRINT DFG TRANSITIONS
    # ============================================================

    print("\n")
    print("=" * 70)
    print("DIRECTLY-FOLLOWS GRAPH")
    print("=" * 70)


    print(
        f"\nNumber of concepts: "
        f"{number_of_cases}"
    )

    print(
        f"Minimum support: "
        f"{min_support}%"
    )

    print(
        f"Minimum transition frequency: "
        f"{min_frequency:.2f}"
    )


    print("\nTransitions:")


    for (source, target), frequency in sorted(
        dfg_filtered.items(),
        key=lambda x: x[1],
        reverse=True
    ):

        relative_frequency = (
            frequency
            / number_of_cases
            * 100
        )

        print(
            f"{source} → {target}: "
            f"{frequency} "
            f"({relative_frequency:.1f}% of number of concepts)"
        )


    # ============================================================
    # 14. VISUALIZE DFG WITH PM4PY
    # ============================================================

    if dfg_filtered:

        pm4py.view_dfg(
            dfg_filtered,
            start_activities_filtered,
            end_activities_filtered,
            rankdir="LR",
            graph_title=(
                f"Directly-Follows Graph für Kategoriengruppe {g} "
                f"(ohne konsekutive Repetition, Filter {min_support}%)"
            )
        )

    else:

        print(
            "\nNo transitions remain after filtering."
        )
```

### Analyse Z.1-1: Cohen's Kappa

```python
def cohens_kappa(n_beobachtungen, n_abweichungen,
                 haeufigkeiten_rater1, haeufigkeiten_rater2):
    """
    Berechnet Cohen's Kappa auf Grundlage der Randhäufigkeiten
    der Kategorien beider Rater.

    Parameter
    ----------
    n_beobachtungen : int
        Gesamtzahl der Beobachtungen.

    n_abweichungen : int
        Anzahl der Beobachtungen, bei denen sich die Rater unterscheiden.

    haeufigkeiten_rater1 : list[int]
        Häufigkeiten der Kategorien bei Rater 1.

    haeufigkeiten_rater2 : list[int]
        Häufigkeiten der Kategorien bei Rater 2.

    Returns
    -------
    float
        Cohen's Kappa.
    """

    # Prüfungen
    if n_beobachtungen <= 0:
        raise ValueError("Die Anzahl der Beobachtungen muss größer als 0 sein.")

    if not 0 <= n_abweichungen <= n_beobachtungen:
        raise ValueError(
            "Die Anzahl der Abweichungen muss zwischen 0 und "
            "der Gesamtzahl der Beobachtungen liegen."
        )

    if len(haeufigkeiten_rater1) != len(haeufigkeiten_rater2):
        raise ValueError(
            "Beide Rater müssen dieselbe Anzahl an Kategorien haben."
        )

    if sum(haeufigkeiten_rater1) != n_beobachtungen:
        raise ValueError(
            "Die Häufigkeiten von Rater 1 müssen sich zur "
            "Gesamtzahl der Beobachtungen summieren."
        )

    if sum(haeufigkeiten_rater2) != n_beobachtungen:
        raise ValueError(
            "Die Häufigkeiten von Rater 2 müssen sich zur "
            "Gesamtzahl der Beobachtungen summieren."
        )

    # Beobachtete Übereinstimmung
    p_o = (n_beobachtungen - n_abweichungen) / n_beobachtungen

    # Relative Häufigkeiten der Kategorien
    p_rater1 = [
        h / n_beobachtungen
        for h in haeufigkeiten_rater1
    ]

    p_rater2 = [
        h / n_beobachtungen
        for h in haeufigkeiten_rater2
    ]

    # Erwartete zufällige Übereinstimmung
    p_e = sum(
        p1 * p2
        for p1, p2 in zip(p_rater1, p_rater2)
    )

    # Cohen's Kappa
    kappa = (p_o - p_e) / (1 - p_e)

    return kappa


# K

n = 471
abweichungen = 6

# -, K-RB, (K-RB + K-RP), K-RF, K-RP
rater1 = [20, 78, 2, 44, 327]
rater2 = [23, 73, 2, 43, 330]

kappa = cohens_kappa(n_beobachtungen=n, n_abweichungen=abweichungen, haeufigkeiten_rater1=rater1, haeufigkeiten_rater2=rater2)

print(f"K - Cohen's Kappa: {kappa:.3f}")


# AE

n = 471
abweichungen = 5

# -, AE-KO, AE-PO, (AE-PO + AE-RO), AE-RO
rater1 = [29, 115, 58, 2, 267]
rater2 = [30, 116, 58, 2, 265]

kappa = cohens_kappa(n_beobachtungen=n, n_abweichungen=abweichungen, haeufigkeiten_rater1=rater1, haeufigkeiten_rater2=rater2)

print(f"AE - Cohen's Kappa: {kappa:.3f}")


# AE

n = 471
abweichungen = 10

rater1 = [0, 14, 2, 24, 2, 0, 27, 10, 108, 4, 3, 1, 77, 3, 7, 124, 17, 45, 2, 0, 1]
rater2 = [1, 14, 2, 23, 2, 1, 28, 10, 103, 4, 3, 2, 75, 3, 8, 126, 18, 44, 2, 1, 1]

kappa = cohens_kappa(n_beobachtungen=n, n_abweichungen=abweichungen, haeufigkeiten_rater1=rater1, haeufigkeiten_rater2=rater2)

print(f"AM - Cohen's Kappa: {kappa:.3f}")
```

```python
import pandas as pd

# Input and output file paths
input_file = "Schritte_final.csv"
output_file = "Schritte_final_ohneDoppel.csv"

# Read CSV with semicolon as delimiter
df = pd.read_csv(input_file, sep=";", dtype=str)

# Create a list for the new rows
new_rows = []

changes = 0

for _, row in df.iterrows():
    value = row['K']

    # Check whether the cell contains a comma
    if pd.notna(value) and "," in value:
        
        # Split at the comma
        parts = value.split(", ")

        # Create one copy of the row for each part
        for part in parts:
            new_row = row.copy()
            new_row['K'] = part.strip()
            new_rows.append(new_row)
        changes += 1
        continue
    else:
        # Keep the row unchanged
        new_rows.append(row)

# Create new DataFrame
new_df = pd.DataFrame(new_rows)
new_rows = []

for _, row in new_df.iterrows():
    value = row['AE']

    # Check whether the cell contains a comma
    if pd.notna(value) and "," in value:
        # Split at the comma
        #parts = value.split(", ")

        # Create one copy of the row for each part
        #for part in parts:
        #    print(part)
        #    new_row = row.copy()
        #    new_row['AE'] = part.strip()
        #    new_rows.append(new_row)
        #    print(new_row)
        changes += 1
        continue
    else:
        # Keep the row unchanged
        new_rows.append(row)

# Create new DataFrame
new_df = pd.DataFrame(new_rows)
new_rows = []

for _, row in new_df.iterrows():
    value = row['AM']

    # Check whether the cell contains a comma
    if pd.notna(value) and "," in value:
        # Split at the comma
        #parts = value.split(",")

        # Create one copy of the row for each part
        #for part in parts:
        #    new_row = row.copy()
        #    new_row['AM'] = part.strip()
        #    new_rows.append(new_row)
        changes += 1
        continue
    else:
        # Keep the row unchanged
        new_rows.append(row)

# Create new DataFrame
new_df = pd.DataFrame(new_rows)

# Save as a new CSV file
new_df.to_csv(output_file, sep=";", index=False)

print(f"Done. Performed {changes} changes. New file saved as: {output_file}")
```
